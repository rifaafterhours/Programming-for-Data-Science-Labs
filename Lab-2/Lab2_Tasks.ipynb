```python
# Create a list containing the marks of 5 courses
marks = [75, 82, 48, 91, 64]

# Display total, average, highest, and lowest mark
total_marks = sum(marks)
average_mark = total_marks / len(marks)
highest_mark = max(marks)
lowest_mark = min(marks)

# Count how many courses the student passed (pass mark = 50)
passed_courses = 0
for mark in marks:
    if mark >= 50:
        passed_courses += 1

print("Course Marks   :", marks)
print("Total Marks    :", total_marks)
print(f"Average Mark   : {average_mark:.2f}")
print("Highest Mark   :", highest_mark)
print("Lowest Mark    :", lowest_mark)
print("Passed Courses :", passed_courses)





# Tuple containing temperatures for Saturday and Sunday (3 times each day)
weekend_temps = (28.5, 31.0, 29.2, 27.8, 33.5, 30.1)

# Display each temperature using a for loop
print("Weekend Temperatures:")
for temp in weekend_temps:
    print(f"Reading: {temp} °C")

# Calculate highest and lowest temperature
highest_temp = weekend_temps[0]
lowest_temp = weekend_temps[0]

for temp in weekend_temps:
    if temp > highest_temp:
        highest_temp = temp
    if temp < lowest_temp:
        lowest_temp = temp

print("\nHighest Temperature:", highest_temp, "°C")
print("Lowest Temperature :", lowest_temp, "°C")





# Ask user for a word or short sentence
text = input("Enter a word or short sentence: ")

vowels = "aeiouAEIOU"
num_characters = 0
num_vowels = 0
num_spaces = 0






# Use a for loop to examine each character
for ch in text:
    num_characters += 1
    if ch in vowels:
        num_vowels += 1
    elif ch == " ":
        num_spaces += 1





print("\nNumber of characters :", num_characters)
print("Number of vowels     :", num_vowels)
print("Number of spaces     :", num_spaces)

# Define the grade calculator function
def calculate_grade(marks):
    if len(marks) == 0:
        return "No marks provided"

    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"







# Call the function using a list of 5 marks
my_marks = [85, 78, 92, 88, 71]
grade = calculate_grade(my_marks)
avg = sum(my_marks) / len(my_marks)

print("Marks List   :", my_marks)
print(f"Average Mark : {avg:.2f}")
print("Final Grade  :", grade)




# List of dictionaries representing 4 products
cart = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]





# Calculate total cost using a for loop
total_cost = 0
for item in cart:
    total_cost += item["price"] * item["quantity"]

# Discount function: 10% discount if total >= 2000
def apply_discount(total):
    if total >= 2000:
        return total * 0.90  # 10% discount
    else:
        return total

final_price = apply_discount(total_cost)

print("Total Cost   :", total_cost, "BDT")
print("Final Price  :", final_price, "BDT")






# Program using a while loop to countdown to zero
number = int(input("Enter a number: "))

while number < 0:
    number = int(input("Please enter a positive number: "))

while number >= 0:
    print(number)
    number -= 1





    # Suggested student dataset from Section 4
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara",  "department": "DS",  "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS",  "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total_marks = 0
passed_count = 0
top_student = students[0]

print("--- Student Records ---")
for s in students:
    print(f"Name: {s['name']:<7} | Marks: {s['marks']}")
    total_marks += s["marks"]
    if s["marks"] >= 50:
        passed_count += 1
    if s["marks"] > top_student["marks"]:
        top_student = s

class_average = total_marks / len(students)

print("\n--- Summary ---")
print(f"Passed Students (>=50) : {passed_count} of {len(students)}")
print(f"Class Average Mark     : {class_average:.2f}")
print(f"Highest Mark Student   : {top_student['name']} ({top_student['marks']} marks)")





# Suggested product dataset from Section 4
products = [
    {"name": "Notebook",  "price": 120, "quantity": 3},
    {"name": "Pen",       "price": 20,  "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse",     "price": 650, "quantity": 1}
]

total_inventory_value = 0
highest_product_name = ""
highest_line_total = -1

print("--- Line Totals ---")
for p in products:
    line_total = p["price"] * p["quantity"]
    total_inventory_value += line_total
    print(f"{p['name']:<10}: {p['price']} * {p['quantity']} = {line_total} BDT")
    
    if line_total > highest_line_total:
        highest_line_total = line_total
        highest_product_name = p["name"]

print("\n--- Summary ---")
print(f"Total Inventory Value  : {total_inventory_value} BDT")
print(f"Highest Line Total Item: {highest_product_name} ({highest_line_total} BDT)")

```
