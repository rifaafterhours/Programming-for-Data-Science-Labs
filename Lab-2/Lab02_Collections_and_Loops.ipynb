```python
# Lab 02 - Problem 1
marks = [75, 82, 48, 91, 64]

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

passed_count = 0
for m in marks:
    if m >= 50:
        passed_count += 1

print("Course Marks   :", marks)
print("Total Marks    :", total)
print(f"Average Mark   : {average:.2f}")
print("Highest Mark   :", highest)
print("Lowest Mark    :", lowest)
print("Passed Courses :", passed_count)

# Lab 02 - Problem 2
weekend_temps = (28.5, 31.0, 29.2, 27.8, 33.5, 30.1)

print("--- Weekend Readings ---")
for temp in weekend_temps:
    print(f"{temp} °C")

# Calculate highest and lowest using loop
highest = weekend_temps[0]
lowest = weekend_temps[0]

for temp in weekend_temps:
    if temp > highest:
        highest = temp
    if temp < lowest:
        lowest = temp

print("\n--- Summary ---")
print("Highest Temperature:", highest, "°C")
print("Lowest Temperature :", lowest, "°C")

# Lab 02 - Problem 3
text = input("Enter a word or short sentence: ")

vowels = "aeiouAEIOU"
char_count = 0
vowel_count = 0
space_count = 0

for ch in text:
    char_count += 1
    if ch in vowels:
        vowel_count += 1
    elif ch == " ":
        space_count += 1

print("\nTotal Characters :", char_count)
print("Total Vowels     :", vowel_count)
print("Total Spaces     :", space_count)

# Lab 02 - Problem 4
def calculate_grade(marks):
    if len(marks) == 0:
        return "No marks provided"
    
    average = sum(marks) / len(marks)
    
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

# Calling the function with a list of five marks
sample_marks = [85, 78, 92, 88, 71]
grade = calculate_grade(sample_marks)
avg = sum(sample_marks) / len(sample_marks)

print("Marks        :", sample_marks)
print(f"Average Mark : {avg:.2f}")
print("Grade        :", grade)

# Lab 02 - Problem 5
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

# Calculate total cost using a loop
total_cost = 0
for item in products:
    total_cost += item["price"] * item["quantity"]

# Discount function
def apply_discount(total):
    if total >= 2000:
        return total * 0.90  # 10% discount
    else:
        return total

final_total = apply_discount(total_cost)

print("Total Cost   :", total_cost, "BDT")
print("Payable Total:", final_total, "BDT")


# Lab 02 - Problem 6
num = int(input("Enter a non-negative integer: "))

while num < 0:
    num = int(input("Invalid! Please enter a NON-NEGATIVE integer: "))

current = num
while current >= 0:
    print(current)
    current -= 1


# Lab 02 - Problem 7
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara",  "department": "DS",  "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS",  "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total_marks = 0
passed_count = 0
top_student = students[0]

print("--- Student Records ---")
for s in students:
    print(f"Name: {s['name']:<7} | Marks: {s['marks']}")
    total_marks += s["marks"]
    if s["marks"] >= 50:
        passed_count += 1
    if s["marks"] > top_student["marks"]:
        top_student = s

class_avg = total_marks / len(students)

print("\n--- Performance Summary ---")
print(f"Passed Students (>=50): {passed_count} of {len(students)}")
print(f"Class Average         : {class_avg:.2f}")
print(f"Top Scorer            : {top_student['name']} ({top_student['marks']} marks)")


# Lab 02 - Problem 8
products = [
    {"name": "Notebook",  "price": 120, "quantity": 3},
    {"name": "Pen",       "price": 20,  "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse",     "price": 650, "quantity": 1}
]

total_inventory_value = 0
highest_product = None
highest_line_total = -1

print("--- Product Line Totals ---")
for p in products:
    line_total = p["price"] * p["quantity"]
    total_inventory_value += line_total
    print(f"{p['name']:<10}: {p['price']} * {p['quantity']} = {line_total} BDT")
    
    if line_total > highest_line_total:
        highest_line_total = line_total
        highest_product = p

print("\n--- Inventory Summary ---")
print(f"Total Inventory Value   : {total_inventory_value} BDT")
print(f"Highest Line Total Item : {highest_product['name']} ({highest_line_total} BDT)")
```
