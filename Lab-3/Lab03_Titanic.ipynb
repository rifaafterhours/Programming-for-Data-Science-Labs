```python
# Load the Titanic dataset
df = pd.read_csv("train.csv")

# Display the first five rows
print("First five rows:")
print(df.head())

# Display the last five rows
print("\nLast five rows:")
print(df.tail())

# Dataset dimensions
print("\nShape:", df.shape)

# Column names
print("\nColumns:", df.columns)

# Data types
print("\nData types:")
print(df.dtypes)

# Descriptive statistics
print("\nDescriptive statistics:")
print(df.describe())
```

```python
# Display the first 10 rows
print(df.head(10))
```

```python
# Find the number of rows and columns
rows, cols = df.shape
print("Number of rows:", rows)
print("Number of columns:", cols)
print("Shape tuple:", df.shape)
```

```python
# Identify all column names
print("Column Names:")
print(list(df.columns))

# Enumerated display
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")
```

```python
# Determine data type of each column
print("Column Data Types:")
print(df.dtypes)
```

```python
# Display descriptive statistics for numerical columns
print(df.describe())
```

```python
# Select one column
print(df["Name"].head())

# Select multiple columns
print(df[["Name", "Age", "Sex", "Survived"]].head())
```

```python
# Select rows using labels
print(df.loc[0:4, ["Name", "Age"]])

# Select rows using integer positions
print(df.iloc[0:5, 0:4])
```

```python
# Display passengers belonging to first class (Pclass == 1)
first_class = df[df["Pclass"] == 1]
print(f"Total First Class Passengers: {len(first_class)}")
print(first_class[["PassengerId", "Name", "Pclass", "Fare", "Survived"]].head())
```

```python
# Find passengers younger than 18
minors = df[df["Age"] < 18]
print(f"Total Passengers younger than 18: {len(minors)}")
print(minors[["PassengerId", "Name", "Age", "Sex", "Pclass", "Survived"]].head())
```

```python
# Find male passengers who survived
male_survivors = df[(df["Sex"] == "male") & (df["Survived"] == 1)]
print(f"Total Male Passengers who survived: {len(male_survivors)}")
print(male_survivors[["PassengerId", "Name", "Age", "Pclass", "Survived"]].head())
```

```python
# Find passengers who paid a fare greater than 100
high_fare = df[df["Fare"] > 100]
print(f"Total Passengers who paid fare > 100: {len(high_fare)}")
print(high_fare[["PassengerId", "Name", "Pclass", "Fare", "Survived"]].head())
```

```python
# Sort passengers by age
print(df.sort_values(by="Age")[["Name", "Age"]].head())

# Sort by fare descending
print(df.sort_values(by="Fare", ascending=False)[["Name", "Fare"]].head())

# Sort by passenger class and fare
print(df.sort_values(by=["Pclass", "Fare"])[["Name", "Pclass", "Fare"]].head())
```

```python
# Basic aggregations
print("Average age:", df["Age"].mean())
print("Median fare:", df["Fare"].median())
print("Max fare:", df["Fare"].max())
print("Min fare:", df["Fare"].min())

# Count passengers by gender
print(df["Sex"].value_counts())

# Grouped aggregations
print(df.groupby("Pclass")["Fare"].mean())
print(df.groupby("Sex")["Survived"].mean())
```

```python
# Display 10 passengers who paid the highest fares
top_10_fares = df.sort_values(by="Fare", ascending=False).head(10)
print(top_10_fares[["PassengerId", "Name", "Pclass", "Fare", "Survived"]])
```

```python
# Count the number of male and female passengers
gender_counts = df["Sex"].value_counts()
print(gender_counts)
print(f"Male count:   {gender_counts['male']}")
print(f"Female count: {gender_counts['female']}")
```

```python
# Calculate the average fare for each passenger class
avg_fare_by_class = df.groupby("Pclass")["Fare"].mean()
print(avg_fare_by_class)
for pclass, fare in avg_fare_by_class.items():
    print(f"Class {pclass}: ${fare:.2f}")
```

```python
# Calculate the survival rate for each passenger class
survival_rate_by_class = df.groupby("Pclass")["Survived"].mean()
print(survival_rate_by_class)
for pclass, rate in survival_rate_by_class.items():
    print(f"Class {pclass} Survival Rate: {rate * 100:.2f}%")
```

```python
total_passengers = len(df)
# Alternatively:
# total_passengers = df.shape[0]
print("Total passengers:", total_passengers)
```

```python
avg_age = df["Age"].mean()
print(f"Average passenger age: {avg_age:.2f} years")
print(f"Median passenger age: {df['Age'].median():.2f} years")
```

```python
class_counts = df["Pclass"].value_counts().sort_index()
top_class = df["Pclass"].value_counts().idxmax()
print(class_counts)
print(f"Highest class: Class {top_class} with {class_counts[top_class]} passengers ({class_counts[top_class]/len(df)*100:.1f}%)")
```

```python
avg_fare_by_gender = df.groupby("Sex")["Fare"].mean()
print(avg_fare_by_gender)
print(f"Female average fare: ${avg_fare_by_gender['female']:.2f}")
print(f"Male average fare:   ${avg_fare_by_gender['male']:.2f}")
```

```python
survival_by_gender = df.groupby("Sex")["Survived"].mean()
print(survival_by_gender)
for sex, rate in survival_by_gender.items():
    print(f"{sex.capitalize()} Survival Rate: {rate * 100:.2f}%")
```

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean()
print(survival_by_class)
for pclass, rate in survival_by_class.items():
    print(f"Class {pclass}: {rate * 100:.2f}%")
```
