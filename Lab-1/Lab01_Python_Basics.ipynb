```python
# Lab 01 - Problem 1
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Sum:", num1 + num2)
print("Difference:", num1 - num2)
print("Product:", num1 * num2)

if num2 != 0:
    print("Quotient:", num1 / num2)
else:
    print("Quotient: Undefined (Cannot divide by zero)")

# Lab 01 - Problem 2
name = input("Enter student name: ")
student_id = input("Enter student ID: ")
age = int(input("Enter age: "))
department = input("Enter department: ")
cgpa = float(input("Enter CGPA: "))

print("\n" + "=" * 35)
print("       STUDENT PROFILE")
print("=" * 35)
print(f"Name       : {name}")
print(f"Student ID : {student_id}")
print(f"Age        : {age}")
print(f"Department : {department}")
print(f"CGPA       : {cgpa:.2f}")
print("=" * 35)

# Lab 01 - Problem 3
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * (9 / 5)) + 32
kelvin = celsius + 273.15

print(f"Temperature in Fahrenheit: {fahrenheit:.2f} °F")
print(f"Temperature in Kelvin    : {kelvin:.2f} K")



# Lab 01 - Problem 4
name = input("Enter student name: ")

marks = []
for i in range(1, 6):
    m = float(input(f"Enter mark for Course {i}: "))
    marks.append(m)

total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

# Count courses with mark >= 50
passed_courses = 0
for m in marks:
    if m >= 50:
        passed_courses += 1

print("\n" + "=" * 40)
print(f"Academic Report for: {name}")
print("=" * 40)
print("Marks List     :", marks)
print(f"Total Marks    : {total}")
print(f"Average Mark   : {average:.2f}")
print(f"Highest Mark   : {highest}")
print(f"Lowest Mark    : {lowest}")
print(f"Passed Courses : {passed_courses} out of {len(marks)}")
print("=" * 40)

# Activity 01 – Variables
student_name = "Rahim"
student_id = "24-12345-1"
age = 21
cgpa = 3.50
department = "Data Science"

print(student_name)
print(student_id)
print(age)
print(cgpa)
print(department)

# Activity 02 – Data Types
name = "Rahim"
age = 21
cgpa = 3.50
is_student = True

print(type(name))
print(type(age))
print(type(cgpa))
print(type(is_student))

# Activity 03 – User Input
name = input("Enter your name: ")
age = input("Enter your age: ")
department = input("Enter your department: ")

print("Hello", name)
print("You are", age, "years old.")
print("You study", department + ".")

# Activity 04 – Basic Arithmetic
a = 25
b = 4

print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Division:", a / b)
print("Remainder:", a % b)
```
